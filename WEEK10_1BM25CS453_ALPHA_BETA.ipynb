```python
import ast
import math


def max_value(state, alpha, beta):
    if not isinstance(state, list):
        return state

    v = -math.inf
    for child in state:
        v = max(v, min_value(child, alpha, beta))
        if v >= beta:
            return v
        alpha = max(alpha, v)
    return v


def min_value(state, alpha, beta):
    if not isinstance(state, list):
        return state

    v = math.inf
    for child in state:
        v = min(v, max_value(child, alpha, beta))
        if v <= alpha:
            return v
        beta = min(beta, v)
    return v


def build_balanced_tree(leaves, branching_factor):
    while len(leaves) > 1:
        leaves = [
            leaves[i : i + branching_factor]
            for i in range(0, len(leaves), branching_factor)
        ]
    return leaves[0]


if __name__ == "__main__":
    print("Choose Tree Type:")
    print("1. Balanced Tree (enter flat leaf values)")
    print("2. Unbalanced / Custom Tree (enter nested list)")
    choice = input("Enter choice (1 or 2): ").strip()

    if choice == "1":
        depth = int(input("Enter depth of the tree: "))
        b = int(
            input(
                "Enter branching factor (e.g., 2 for binary, 3 for ternary): "
            )
        )
        expected_leaves = b**depth

        print(f"Enter {expected_leaves} leaf values separated by spaces:")
        leaves = list(map(int, input().split()))

        if len(leaves) != expected_leaves:
            print(
                f"Error: Expected {expected_leaves} values, got {len(leaves)}."
            )
            exit()

        game_tree = build_balanced_tree(leaves, b)

    elif choice == "2":
        print("Enter the tree as a nested list using square brackets []:")
        user_input = input("Tree: ")
        game_tree = ast.literal_eval(user_input)

    else:
        print("Invalid choice.")
        exit()

    optimal_value = max_value(game_tree, -math.inf, math.inf)
    print(f"\nOptimal Value: {optimal_value}")
```
