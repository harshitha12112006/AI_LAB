```python
def get_boolean_input(prompt):
    """Helper function to ensure the user inputs valid true/false values."""
    while True:
        val = input(prompt).strip().lower()
        if val in ['true', 't', '1']:
            return True
        elif val in ['false', 'f', '0']:
            return False
        print("Invalid input! Please enter 'true' or 'false'.")

def manual_truth_table():
    while True:
        try:
            num_entries = int(input("How many entries (rows) do you want to enter? "))
            if num_entries > 0:
                break
            print("Please enter a number greater than 0.")
        except ValueError:
            print("Invalid number! Please enter an integer.")

    rows_data = []

    for i in range(1, num_entries + 1):
        print(f"\n--- Entry #{i} ---")
        A = get_boolean_input("Enter value for A (true/false): ")
        B = get_boolean_input("Enter value for B (true/false): ")
        C = get_boolean_input("Enter value for C (true/false): ")


        A_or_C = A or C
        B_or_not_C = B or (not C)
        kb = A_or_C and B_or_not_C
        alpha = A or B


        rows_data.append((A, B, C, A_or_C, B_or_not_C, kb, alpha))


    header = f"{'A':<8} | {'B':<8} | {'C':<8} | {'A V C':<8} | {'B V ~C':<8} | {'KB':<8} | {'alpha':<8}"
    print("\n" + header)
    print("-" * len(header))

    entails = True

    for A, B, C, A_or_C, B_or_not_C, kb, alpha in rows_data:
        if kb and not alpha:
            entails = False

        str_A = str(A).lower()
        str_B = str(B).lower()
        str_C = str(C).lower()
        str_AvC = str(A_or_C).lower()
        str_BvnotC = str(B_or_not_C).lower()
        str_kb = str(kb).lower()
        str_alpha = str(alpha).lower()

        print(f"{str_A:<8} | {str_B:<8} | {str_C:<8} | {str_AvC:<8} | {str_BvnotC:<8} | {str_kb:<8} | {str_alpha:<8}")

    print("-" * len(header))
    print(f"Conclusion: KB |= alpha is {entails}\n")

if __name__ == "__main__":
    manual_truth_table()
```
