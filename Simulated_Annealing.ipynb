```python

import random
import math

N = 8

def calculate_cost(board):
    conflicts = 0

    for c1 in range(N):
        for c2 in range(c1 + 1, N):
            if (board[c1] == board[c2] or
                abs(board[c1] - board[c2]) == abs(c1 - c2)):
                conflicts += 1

    return conflicts

def generate_neighbor(board):
    new_board = board.copy()

    col = random.randint(0, N - 1)
    old_row = new_board[col]

    new_row = random.randint(0, N - 2)
    if new_row >= old_row:
        new_row += 1

    new_board[col] = new_row
    return new_board


def simulated_annealing():
    initial_temperature = 10.0
    cooling_factor = 0.995
    max_iterations = 10000
    max_restarts = 100

    for restart in range(max_restarts):
        board = [random.randint(0, N - 1) for _ in range(N)]
        current_cost = calculate_cost(board)

        temperature = initial_temperature
        best_board = board.copy()
        best_cost = current_cost

        for step in range(max_iterations):
            if current_cost == 0:
                return board, current_cost, restart + 1, step

            neighbor = generate_neighbor(board)
            new_cost = calculate_cost(neighbor)

            delta = new_cost - current_cost

            if delta < 0:
                accept = True
            else:
                probability = math.exp(-delta / temperature)
                accept = random.random() < probability

            if accept:
                board = neighbor
                current_cost = new_cost

            if current_cost < best_cost:
                best_board = board.copy()
                best_cost = current_cost

            temperature *= cooling_factor

    return best_board, best_cost, max_restarts, max_iterations


def display_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


solution, cost, restarts, steps = simulated_annealing()

print("8-Queens Problem using Simulated Annealing")
print("Solution:", solution)
print("Final Cost:", cost)
print("Restarts:", restarts)
print("Iterations:", steps)

if cost == 0:
    print("\nSolution found successfully!")
    display_board(solution)
else:
    print("\nNo solution found within the iteration limit.")
```
