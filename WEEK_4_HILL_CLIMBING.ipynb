```python
def calculate_cost(state):
    """
    Calculates the number of attacking pairs of queens.
    Since each queen is in a unique row and column (permutation),
    we only need to check for diagonal attacks.
    """
    cost = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Diagonal attack condition: |col_i - col_j| == |row_i - row_j|
            if abs(i - j) == abs(state[i] - state[j]):
                cost += 1
    return cost

def get_best_neighbor(state):
    """
    Generates all neighbors by swapping pairs of queens (columns).
    Returns the best neighbor based on the minimum cost and the tie-breaking rule
    (smallest subscripts/column numbers first).
    """
    n = len(state)

    best_neighbor = None
    best_cost = float('inf')
    best_pair = (float('inf'), float('inf'))

    # Iterate through all pairs to swap, ordered lexicographically (i, j)
    # to naturally respect the tie-breaking condition
    for i in range(n):
        for j in range(i + 1, n):
            # Create a neighbor by swapping row positions of column i and column j
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbor = tuple(neighbor)

            cost = calculate_cost(neighbor)

            # Tie-breaking logic: select the lower cost. If costs are equal,
            # choose the one with the smaller (i, j) indices.
            if cost < best_cost:
                best_cost = cost
                best_neighbor = neighbor
                best_pair = (i, j)
            elif cost == best_cost:
                if (i, j) < best_pair:
                    best_neighbor = neighbor
                    best_pair = (i, j)

    return best_neighbor, best_cost

def print_board(state):
    """
    Prints a visual representation of the chessboard layout based on the state.
    """
    n = len(state)
    print("\nFinal Chessboard Layout:")
    # Print column headers
    print("      " + " ".join(f"Col {c}" for c in range(n)))

    # Print grid row by row
    for r in range(n):
        row_str = f"Row {r} "
        for c in range(n):
            if state[c] == r:
                row_str += "[ Q ] "
            else:
                row_str += "[   ] "
        print(row_str)

def hill_climbing_4_queens(initial_state):
    """
    Performs the Hill-Climbing search algorithm to solve the 4-Queens problem.
    """
    current_state = tuple(initial_state)
    current_cost = calculate_cost(current_state)

    print(f"Initial State: {current_state} | Cost: {current_cost}")

    step = 1
    while True:
        neighbor, neighbor_cost = get_best_neighbor(current_state)

        # In hill climbing (minimizing cost), if the best neighbor does not
        # improve the cost (strictly less), we terminate at a local or global optimum.
        if neighbor_cost >= current_cost:
            print(f"\nTerminated at state: {current_state} with Cost: {current_cost}")
            if current_cost == 0:
                print("Success: Global optimum found!")
            else:
                print("Failed: Stuck in a local optimum / plateau.")

            # Call the board visualization function here
            print_board(current_state)
            return current_state

        print(f"Step {step}: Moved to {neighbor} | Cost: {neighbor_cost}")
        current_state = neighbor
        current_cost = neighbor_cost
        step += 1

# Define the initial state provided in the lab slide Example:
# x0=3, x1=1, x2=2, x3=0
initial_state = [0, 1, 2, 3]
hill_climbing_4_queens(initial_state)
```
