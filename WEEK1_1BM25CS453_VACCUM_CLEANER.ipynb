```python
# Simple Reflex Vacuum Cleaner Agent

room = {
    "A": "Dirty",
    "B": "Dirty"
}

position = "A"


def display_state():
    print("Room A:", room["A"])
    print("Room B:", room["B"])
    print("Vacuum Position:", position)
    print()


print("SIMPLE REFLEX VACUUM CLEANER AGENT")
print("-----------------------------------")
print("Harshitha B J (1BM25CS453)")

while room["A"] == "Dirty" or room["B"] == "Dirty":

    display_state()

    # Percept = current room condition

    if room[position] == "Dirty":
        print("Action: SUCK")
        room[position] = "Clean"

    elif position == "A":
        print("Action: MOVE RIGHT")
        position = "B"

    else:
        print("Action: MOVE LEFT")
        position = "A"

print("\nFinal State:")
display_state()

print("Goal achieved: Both rooms are clean.")

```
