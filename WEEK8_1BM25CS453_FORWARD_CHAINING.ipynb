```python

def print_facts(fact_list):
    print("\nFacts:")
    for fact in fact_list:
        print(fact)


def print_rules(rule_list):
    print("\nRules:")
    for rule in rule_list:
        premise, conclusion = rule
        print(f"if {premise}, then {conclusion}")


def is_condition_satisfied(premise, facts):
    premises = premise.split(' and ')

    for cond in premises:
        if cond not in facts:
            return False

    return True

def forward_chaining(initial_facts, rules, query=None):

    facts = initial_facts.copy()

    print("\nInference Derivation Trace:")

    if query and query in facts:
        print(f"\n[SUCCESS] Goal '{query}' is already known!")
        return facts

    while True:

        newly_derived_facts = []

        for rule in rules:

            premise, conclusion = rule

            if (is_condition_satisfied(premise, facts)
                    and conclusion not in facts):

                newly_derived_facts.append(conclusion)
                facts.append(conclusion)

                print(f" -> Derived: {conclusion}")

                if query and conclusion == query:
                    print(
                        f"\n[SUCCESS] Goal '{query}' successfully proved!"
                    )
                    return facts

        if not newly_derived_facts:
            break

    if query and query not in facts:
        print(f"\n[FAILED] Could not prove the goal '{query}'.")

    return facts

if __name__ == "__main__":

    facts = []
    rules = []

    while True:

        print("\n-----------------------------------------")
        print("[1] Add facts")
        print("[2] Add rules")
        print("[3] Done Adding")
        print("[4] Generate, display, and prove a query")
        print("[5] Exit")

        choice = input("\nEnter your choice: ")

        if choice == '5':
            break

        elif choice == '1':

            ctr_fact = 1
            print("\nAdding Fact(s)...")

            while True:

                fact_input = input(f"Fact {ctr_fact}: ")

                if not fact_input.strip():
                    print("Invalid input")
                    continue

                if fact_input == '3':
                    break

                facts.append(fact_input.strip())
                ctr_fact += 1

            print("\nFact(s) added.\n")
            print_facts(facts)
            print_rules(rules)

        elif choice == '2':

            ctr_rule = 1
            print("\nAdding Rule(s)...")

            while True:

                rule_input = input(f"Rule {ctr_rule}: ")

                if rule_input == '3':
                    break

                if "if " in rule_input and ", then " in rule_input:

                    premise, conclusion = rule_input.split(
                        ", then ", 1
                    )

                    premise = premise.replace("if ", "", 1).strip()
                    conclusion = conclusion.strip()

                    rules.append((premise, conclusion))
                    ctr_rule += 1

                else:
                    print(
                        "Invalid rule format. "
                        "[if {premise}, then {conclusion}]"
                    )

            print("\nRule(s) added.\n")
            print_facts(facts)
            print_rules(rules)


        elif choice == '3':

            print(
                "\nDone adding. "
                "You can now choose option [4] to run the query."
            )


        elif choice == '4':

            query_input = input(
                "\nEnter the query/goal you want to prove: "
            ).strip()

            updated_facts = forward_chaining(
                facts, rules, query=query_input
            )

            print_facts(updated_facts)

        else:
            print("Invalid choice.")
```
